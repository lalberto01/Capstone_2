## 1. Create a list of all transactions from January 2024, sorted by sale amount from highest to lowest. Which transactions had a sale amount greater than $500?

```python
# transactions from Jan 2024
january_2024 = store_sales[
    store_sales["Transaction Date"].dt.strftime("%Y-%m") == "2024-01"
]
```

```python
# Sort from highest to lowest
january_2024 = january_2024.sort_values(
    by="Sale Amount",
    ascending=False
)

january_2024
```

```python
# Transactions higher then $500
high_sales = january_2024[
    january_2024["Sale Amount"] > 500
]

high_sales
```
