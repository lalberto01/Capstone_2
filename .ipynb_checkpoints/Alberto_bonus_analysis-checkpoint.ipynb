# Capstone Bonus Questions

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
store_sales = pd.read_csv("StoreSales.csv")
store_detail = pd.read_csv("StoreDetail.csv")
products = pd.read_csv("Products.csv")
product_categories = pd.read_csv("ProductCategories.csv")
customer_list = pd.read_csv("customer_list.csv", sep="|")
```

```python
store_sales["Transaction Date"] = pd.to_datetime(store_sales["Transaction Date"])
```

```python
sales_data = store_sales.merge(store_detail, on="Store ID", how="inner")
```

```python
south_sales = sales_data[
    sales_data["State"].isin(["Florida", "Texas"])
].copy()
```

## 1. Create a list of all transactions from January 2024, sorted by sale amount from highest to lowest. Which transactions had a sale amount greater than $500?

```python
# transactions from Jan 2024
january_2024 = store_sales[
    store_sales["Transaction Date"].dt.strftime("%Y-%m") == "2024-01"
]
```

```python
# Sort from highest to lowest
january_2024 = january_2024.sort_values(
    by="Sale Amount",
    ascending=False
)

january_2024
```

```python
# Transactions higher then $500
high_sales = january_2024[
    january_2024["Sale Amount"] > 500
]

high_sales
```

# 2. Find all products whose product number begins with 10525. What category and subcategory do they belong to?

```python
# Product number begins with 10525
products_10525 = products[
    products["Prod Num"].astype(str).str.startswith("10525")
]

products_10525
```

```python
# Merge with category information
products_10525 = products_10525.merge(
    product_categories,
    on=["CategoryID", "SubcategoryID"],
    how="inner"
)

products_10525[
    ["Prod Num", "Category", "Subcategory"]
]
```

# 3. Which stores in your assigned territories had at least one transaction with a sale amount over $1,000? How many such transactions did each store have?

```python
# Find transactions greater than $1,000
high_sales = south_sales[
    south_sales["Sale Amount"] > 1000
]

high_sales
```

```python
# Count transactions over $1,000 by store
high_sales_summary = high_sales.groupby(
    ["State", "Store ID", "Store Location"]
).size().reset_index(name="Transactions Over 1000")

high_sales_summary
```
