## Monthly Revenue Analysis

This will calculate the in store monthly sales revenue for Flordia and Texas from the data. Flordia generated the highest in Oct 2025 with $188,215.96, while Texas numbers slowed down. Focus on marketing during the end of the year. Since numbers start to lower, at least to keep it at a steady rate. Until Januaray when sales may go up due back to school. 

```python
# Convert Transaction Date column into date format
store_sales["Transaction Date"] = pd.to_datetime(
    store_sales["Transaction Date"]
)
```

```python
# Only 2025 transactions
sales_2025 = store_sales[
    store_sales["Transaction Date"].dt.year == 2025
].copy()
```

```python
# Merge 2025 sales data with store details
sales_data = sales_2025.merge(
    store_detail,
    on="Store ID",
    how="inner"
)
```

```python
# Keep only Florida and Texas
south_sales = sales_data[
    sales_data["State"].isin(["Florida", "Texas"])
].copy()
```

```python
# Create Month column
south_sales["Month"] = south_sales[
    "Transaction Date"
].dt.to_period("M")
```

```python
# Calculate monthly revenue by state
monthly_revenue = south_sales.groupby(
    ["State", "Month"]
)["Sale Amount"].sum().reset_index()

monthly_revenue.sort_values(["Month", "State"])
```

```python
# Calculate monthly revenue totals
monthly_chart = south_sales.groupby(
    ["State", "Month"]
)["Sale Amount"].sum().reset_index()

# Florida monthly revenue
florida_chart = monthly_chart[
    monthly_chart["State"] == "Florida"
]

# Texas monthly revenue
texas_chart = monthly_chart[
    monthly_chart["State"] == "Texas"
]

plt.figure(figsize=(10,5))

plt.plot(
    florida_chart["Month"].astype(str),
    florida_chart["Sale Amount"],
    marker="o",
    label="Florida",
    color="orange"
)

plt.plot(
    texas_chart["Month"].astype(str),
    texas_chart["Sale Amount"],
    marker="o",
    label="Texas",
    color="green"
)

plt.xticks(rotation=45)

plt.title("2025 Monthly Revenue in Florida and Texas")
plt.xlabel("Month")
plt.ylabel("Total Revenue")

plt.legend()

plt.show()
```

# Store Sales 

This section ranks each store in Florida and Texas by total sales revenue. Flordia store 724 and Texas store 905 were pretty close in revenue store 724 at the top by a $27,446.80 difference. Focusing on marketing during the highest peak of sales can help in the future revenue. Most of the time you have a high demand of customers at these locations. Important for future marketing investment and customer retentation. Pushing a great customer experience will have them return. 

```python
# Convert Transaction Date column into date format
store_sales["Transaction Date"] = pd.to_datetime(
    store_sales["Transaction Date"]
)
```

```python
# Only 2025 transactions
sales_2025 = store_sales[
    store_sales["Transaction Date"].dt.year == 2025
].copy()
```

```python
# Merge 2025 sales with store details
sales_data = sales_2025.merge(
    store_detail,
    on="Store ID",
    how="inner"
)
```

```python
# Keep only Florida and Texas
south_sales = sales_data[
    sales_data["State"].isin(["Florida", "Texas"])
].copy()
```

```python
# Calculate total store revenue
store_performance = south_sales.groupby(
    ["State", "Store ID", "Store Location"]
)["Sale Amount"].sum().reset_index()

store_performance
```

```python
# Show top stores in each state
top_stores = store_performance.groupby(
    "State"
).head(5)

top_stores
```

```python
# Top 5 stores from each state
top_store_chart = store_performance.groupby(
    "State"
).head(5)

# Separate Florida and Texas stores
florida_stores = top_store_chart[
    top_store_chart["State"] == "Florida"
]

texas_stores = top_store_chart[
    top_store_chart["State"] == "Texas"
]

plt.figure(figsize=(10,5))

plt.plot(
    florida_stores["Store Location"],
    florida_stores["Sale Amount"],
    marker="o",
    label="Florida",
    color="orange"
)

plt.plot(
    texas_stores["Store Location"],
    texas_stores["Sale Amount"],
    marker="o",
    label="Texas",
    color="green"
)

plt.xticks(rotation=45)

plt.title("2025 Top Store Sales in Florida and Texas")
plt.xlabel("Store Location")
plt.ylabel("Total Revenue")

plt.legend()

plt.show()
```

# Product Category Analysis
The number of transactions and total revenue by product category for Florida and Texas. Texas and Flordia highest catergory for the last 6 months in Oct 2025 was technology. Which is great but need to focus on other low categories as well. Such as Stationery and Supplies for the year of 2025 was only $49,432.90. Apparel and Merchandise also have the potential to grow more as that is one of the top categories during peak months. 


```python
# Only 2025 transactions
sales_2025 = store_sales[
    store_sales["Transaction Date"].dt.year == 2025
].copy()
```

```python
# Create month column
sales_2025["Month"] = sales_2025[
    "Transaction Date"
].dt.to_period("M")
```

```python
# Merge 2025 sales with store details
sales_data = sales_2025.merge(
    store_detail,
    on="Store ID",
    how="inner"
)
```

```python
# Keep only Florida and Texas
south_sales = sales_data[
    sales_data["State"].isin(["Florida", "Texas"])
].copy()
```

```python
# Merge with products data
sales_products = south_sales.merge(
    products,
    on="Prod Num",
    how="inner"
)
```

```python
# Merge with product categories
sales_products = sales_products.merge(
    product_categories,
    on=["CategoryID", "SubcategoryID"],
    how="inner"
)
```

```python
# Count monthly transactions by category
category_transactions = sales_products.groupby(
    ["State", "Month", "Category"]
).size().reset_index(name="Transactions")

category_transactions.sort_values(["Month", "State"])
```

```python
# Texas last 6 months transactions by category
texas_chart = sales_products[
    (sales_products["State"] == "Texas") &
    (sales_products["Transaction Date"].dt.month >= 7)
]

texas_chart = texas_chart.groupby(
    ["Month", "Category"]
).size().reset_index(name="Transactions")

plt.figure(figsize=(12,5))

for category in texas_chart["Category"].unique():
    
    category_data = texas_chart[
        texas_chart["Category"] == category
    ]
    
    plt.plot(
        category_data["Month"].astype(str),
        category_data["Transactions"],
        marker="o",
        label=category
    )

plt.xticks(rotation=45)

plt.title("Texas Product Category Trends (Last 6 Months of 2025)")
plt.xlabel("Month")
plt.ylabel("Transactions")

plt.legend()

plt.show()
```

```python
# Florida last 6 months transactions by category
florida_chart = sales_products[
    (sales_products["State"] == "Florida") &
    (sales_products["Transaction Date"].dt.month >= 7)
]

florida_chart = florida_chart.groupby(
    ["Month", "Category"]
).size().reset_index(name="Transactions")

plt.figure(figsize=(12,5))

for category in florida_chart["Category"].unique():
    
    category_data = florida_chart[
        florida_chart["Category"] == category
    ]
    
    plt.plot(
        category_data["Month"].astype(str),
        category_data["Transactions"],
        marker="o",
        label=category
    )

plt.xticks(rotation=45)

plt.title("Florida Product Category Trends (Last 6 Months of 2025)")
plt.xlabel("Month")
plt.ylabel("Transactions")

plt.legend()

plt.show()
```

```python
# Calculate monthly revenue by category
category_revenue = sales_products.groupby(
    ["State", "Month", "Category"]
)["Sale Amount"].sum().reset_index()

category_revenue.sort_values(["Month", "State"])
```

```python
# Calculate total 2025 revenue by product category
category_sales = sales_products.groupby(
    "Category"
)["Sale Amount"].sum().reset_index()

category_sales = category_sales.sort_values(
    by="Sale Amount",
    ascending=False
)

category_sales
```

```python
# Bar chart for 2025 product category sales
plt.figure(figsize=(10,5))

plt.bar(
    category_sales["Category"],
    category_sales["Sale Amount"],
    color="pink"
)

plt.xticks(rotation=45)

plt.title("2025 Sales by Product Category in Florida and Texas")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")

plt.show()
```
